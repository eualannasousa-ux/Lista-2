```python
horas = int(input("Horas: "))

if horas <= 2:
    valor = 8.00
elif horas <= 5:
    valor = 8.00 + (horas - 2) * 3.00
else:
    valor = 17.00 + (horas - 5) * 2.00

print(f"Valor a pagar: R$ {valor:.2f}")
```

```python
preco = 25
total = 0
while total < preco:
    v = int(input("Insira um valor: "))
    total += v

print("Total inserido:", total)
print("Troco:", total - preco)
```

```python
maior = 0
pos_maior = 0
for i in range(1, 9):
    n = int(input(f"Digite o {i}º número: "))
    if i == 1 or n > maior:
        maior = n
        pos_maior = i

print("Maior:", maior, "Posição:", pos_maior)
```

```python
maior = 0
segundo = 0
for i in range(8):
    n = int(input("Número: "))
    if i == 0 or n > maior:
        segundo = maior
        maior = n
    elif i == 1 or n > segundo:
        # ajusta para o segundo caso na 2ª leitura
        if n < maior and n > segundo:
            segundo = n

print("Maior:", maior)
print("Segundo maior:", segundo)
```

```python
anterior = 0
repetiu = False
for i in range(8):
    n = int(input("Número: "))
    if i > 0 and n == anterior:
        repetiu = True
    anterior = n

if repetiu:
    print("existem valores consecutivos iguais")
else:
    print("não existem valores consecutivos iguais")
```

```python
saldo = 500.0
for i in range(6):
    tipo = int(input("Tipo (1-depósito 2-saque): "))
    valor = float(input("Valor: "))
    if tipo == 1:
        saldo += valor
    elif tipo == 2:
        if valor <= saldo:
            saldo -= valor
        else:
            print("Saldo insuficiente")

print(f"Saldo final: R$ {saldo:.2f}")
```

```python
n = int(input("Digite um inteiro >1: "))
eh_primo = True
for i in range(2, n):
    if n % i == 0:
        eh_primo = False
        break

if eh_primo:
    print("É primo")
else:
    print("Não é primo")
```

```python
n = int(input("N: "))
qtd = 0
for i in range(1, n+1):
    if n % i == 0:
        print(i)
        qtd += 1
print("Quantidade de divisores:", qtd)
```

```python
a = int(input("A: "))
b = int(input("B: "))
total = 0
for n in range(a, b+1):
    if n > 1:
        eh_primo = True
        for i in range(2, n):
            if n % i == 0:
                eh_primo = False
                break
        if eh_primo:
            print(n)
            total += 1
print("Encontrados:", total)
```

```python
secreto = 37
for tentativa in range(1, 6):
    palpite = int(input(f"Tentativa {tentativa}: "))
    if palpite == secreto:
        print("Acertou!")
        break
    elif palpite < secreto:
        print("O secreto é maior")
    else:
        print("O secreto é menor")
else:
    print("Acabaram as tentativas")
```

```python
saldo = 200.0
qtd_entradas = 0
qtd_retiradas_ok = 0
qtd_retiradas_rec = 0
total_entrou = 0.0
total_saiu = 0.0

while True:
    codigo = int(input("Código (1-entrada 2-retirada 0-sair): "))
    if codigo == 0:
        break
    if codigo == 1:
        valor = float(input("Valor entrada: "))
        saldo += valor
        total_entrou += valor
        qtd_entradas += 1
    elif codigo == 2:
        valor = float(input("Valor retirada: "))
        if valor <= saldo:
            saldo -= valor
            total_saiu += valor
            qtd_retiradas_ok += 1
        else:
            print("Saldo insuficiente")
            qtd_retiradas_rec += 1

print(f"Saldo final: {saldo}")
print("Entradas:", qtd_entradas, "total:", total_entrou)
print("Retiradas ok:", qtd_retiradas_ok, "recusadas:", qtd_retiradas_rec, "total saiu:", total_saiu)
```
